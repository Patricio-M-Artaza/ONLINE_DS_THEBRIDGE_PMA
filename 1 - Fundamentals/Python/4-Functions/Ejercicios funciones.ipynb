## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
# funcion que convierte numeros del 1 al 7 en dias de la semana. un argumento numerico y una salida tipo str.
dias_semana = {1: "Lunes", 2: "Martes", 3: "Miercoles", 4: "Jueves", 5: "Viernes", 6: "Sabado", 7: "Domingo"}

def semana(numero):
    if numero in dias_semana:
        valor = dias_semana.get(numero)
        return valor
    else:
        return "No esta el dia de la semana."

print(semana(7))
    
````

## Ejercicio 3
Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python
# funcion que compare dos numeros. Tiene dos argumentos y 3 salidas posibles: iguales, el primero > segundo, segungo > el primero.

def compare(a, b):
    if a == b:
        return f"{a} y {b} son iguales"
    elif a > b:
        return f"El primer argumento {a} es mayor que {b}"
    else:
        return f"El segundo argumento {b} es mayor que el primero {a}"

print(compare(11,121))
````

## Ejercicio 4

````python
# funcion que cuente letras. Primer argumento se introduce un texto y el seguno se introduce una letra a contar, devuelve un entero con el numero de veces que aparece esa letra
# mayusculas o minuscula.

def contador_letras(palabra, letra):
    palabra_minus = palabra.lower()
    return palabra_minus.count(letra.lower())

        


print(contador_letras("HOLA", "A"))


````
